```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression as SkLogisticRegression

from mydl.models import LogisticRegression
```

```python
data = load_breast_cancer()

X = data.data
Y = data.target
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train = X_train.T
X_test = X_test.T

Y_train = Y_train.reshape(1, -1)
Y_test = Y_test.reshape(1, -1)
```

```python
model = LogisticRegression(
    learning_rate=0.01,
    epochs=5000,
)

model.fit(X_train, Y_train)
```

```python
accuracy = model.score(X_test, Y_test)

print(f"My Model Accuracy: {accuracy:.4f}")
```

```python
sk_model = SkLogisticRegression(max_iter=5000)

sk_model.fit(X_train.T, Y_train.ravel())
```

```python
sk_accuracy = sk_model.score(
    X_test.T,
    Y_test.ravel()
)

print(f"Scikit-Learn Accuracy: {sk_accuracy:.4f}")
```

```python
plt.figure(figsize=(8,5))

plt.plot(model.cost_history)

plt.xlabel("Epoch")

plt.ylabel("Cost")

plt.title("Training Loss")

plt.grid(True)

plt.show()
```
